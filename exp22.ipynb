Reference Resolution

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

sentence = "Ravi went to the library. He borrowed a book."

sentences = sentence.split(".")

previous_nouns = []

for sent in sentences:

    words = word_tokenize(sent)

    print("\nSentence:", sent)

    for word in words:

        if word.lower() in ["he", "she", "it", "they"]:
            if previous_nouns:
                print(
                    word,
                    "refers to",
                    previous_nouns[-1]
                )

        elif word[0].isupper() and word.lower() not in ["He", "She"]:
            previous_nouns.append(word)
```
