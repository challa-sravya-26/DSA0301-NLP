Porter Stemmer

```python
!pip install nltk -q

import nltk
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "running",
    "easily",
    "fairness"
]

print("Word -> Stem")
print("------------")

for word in words:
    print(word, "->", stemmer.stem(word))
```
