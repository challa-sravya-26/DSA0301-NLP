Earley Parser

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"]]
}

sentence = "the cat sees the dog"
words = sentence.split()


def is_nonterminal(symbol):
    return symbol in grammar


def earley_parse(words):

    n = len(words)
    chart = [set() for _ in range(n + 1)]

    # State format:
    # (lhs, tuple(rhs), dot, origin)

    chart[0].add(("S'", ("S",), 0, 0))

    for i in range(n + 1):

        changed = True

        while changed:
            changed = False

            for state in list(chart[i]):

                lhs, rhs, dot, origin = state

                # Complete
                if dot == len(rhs):

                    for old_state in list(chart[origin]):

                        old_lhs, old_rhs, old_dot, old_origin = old_state

                        if (
                            old_dot < len(old_rhs)
                            and old_rhs[old_dot] == lhs
                        ):
                            new_state = (
                                old_lhs,
                                old_rhs,
                                old_dot + 1,
                                old_origin
                            )

                            if new_state not in chart[i]:
                                chart[i].add(new_state)
                                changed = True

                else:
                    next_symbol = rhs[dot]

                    # Predict
                    if is_nonterminal(next_symbol):

                        for production in grammar[next_symbol]:

                            new_state = (
                                next_symbol,
                                tuple(production),
                                0,
                                i
                            )

                            if new_state not in chart[i]:
                                chart[i].add(new_state)
                                changed = True

        # Scan
        if i < n:

            for state in list(chart[i]):

                lhs, rhs, dot, origin = state

                if dot < len(rhs):

                    next_symbol = rhs[dot]

                    if (
                        not is_nonterminal(next_symbol)
                        and next_symbol == words[i]
                    ):

                        chart[i + 1].add(
                            (lhs, rhs, dot + 1, origin)
                        )

    final_state = ("S'", ("S",), 1, 0)

    return final_state in chart[n]


if earley_parse(words):
    print("Sentence is accepted by Earley Parser.")
else:
    print("Sentence is rejected by Earley Parser.")
```
