Agreement Checking

```python
singular_subjects = ["he", "she", "it", "boy", "girl", "cat"]
plural_subjects = ["they", "we", "boys", "girls", "cats"]

singular_verbs = ["is", "was", "runs", "plays", "eats"]
plural_verbs = ["are", "were", "run", "play", "eat"]


def check_agreement(sentence):

    words = sentence.lower().split()

    if len(words) < 2:
        return False

    subject = words[0]
    verb = words[1]

    if subject in singular_subjects and verb in singular_verbs:
        return True

    if subject in plural_subjects and verb in plural_verbs:
        return True

    return False


sentences = [
    "he runs",
    "they run",
    "she are",
    "boys plays"
]

for sentence in sentences:

    if check_agreement(sentence):
        print(sentence, "-> Correct Agreement")
    else:
        print(sentence, "-> Incorrect Agreement")
```
