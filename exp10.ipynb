Transformation-Based Tagging

```python
def initial_tag(word):
    if word.endswith("ing"):
        return "NN"
    elif word.endswith("ed"):
        return "NN"
    else:
        return "NN"


def transformation_based_tagger(words):

    tagged = [(word, initial_tag(word)) for word in words]

    for i, (word, tag) in enumerate(tagged):

        # Transformation rule 1:
        # Words ending in ing are verbs
        if word.endswith("ing"):
            tagged[i] = (word, "VBG")

        # Transformation rule 2:
        # Words ending in ed are past tense verbs
        elif word.endswith("ed"):
            tagged[i] = (word, "VBD")

        # Transformation rule 3:
        # Words ending in ly are adverbs
        elif word.endswith("ly"):
            tagged[i] = (word, "RB")

    return tagged


sentence = "The boy played happily"

words = sentence.split()

result = transformation_based_tagger(words)

print("Transformation-Based Tags:")

for word, tag in result:
    print(word, "->", tag)
```
