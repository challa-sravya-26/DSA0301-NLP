Backoff model

```python
from collections import Counter

# Training corpus
text = "I love natural language processing I love python"

# Tokenize
tokens = text.lower().split()

# Count unigrams and bigrams
unigrams = Counter(tokens)
bigrams = Counter(zip(tokens[:-1], tokens[1:]))

# Total number of words
total_words = len(tokens)

# Backoff model
def backoff_probability(prev_word, curr_word):

    # Use bigram probability if available
    if bigrams[(prev_word, curr_word)] > 0:
        return bigrams[(prev_word, curr_word)] / unigrams[prev_word]

    # Otherwise use unigram probability
    else:
        return unigrams[curr_word] / total_words if curr_word in unigrams else 0

# Examples
print("Bigram Probability P(love|i):", backoff_probability("i", "love"))
print("Backoff Probability P(java|love):", backoff_probability("love", "java"))
```
