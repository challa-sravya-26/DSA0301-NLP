Rule-Based POS Tagging Using Regular Expressions

```python
import re

def rule_based_tagger(sentence):
    words = sentence.split()
    tagged_words = []

    for word in words:

        if re.match(r'.*ing$', word):
            tag = "VBG"

        elif re.match(r'.*ed$', word):
            tag = "VBD"

        elif re.match(r'.*ly$', word):
            tag = "RB"

        elif re.match(r'.*ness$', word):
            tag = "NN"

        elif word.lower() in ["the", "a", "an"]:
            tag = "DT"

        elif word.lower() in ["is", "am", "are", "was", "were"]:
            tag = "VB"

        elif word[0].isupper():
            tag = "NNP"

        else:
            tag = "NN"

        tagged_words.append((word, tag))

    return tagged_words


sentence = "The student is studying carefully"

print("Rule-Based POS Tagging:")
for word, tag in rule_based_tagger(sentence):
    print(word, "->", tag)
```
