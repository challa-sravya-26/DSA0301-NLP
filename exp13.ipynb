Generate a Parse Tree

```python
import nltk

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the'
N -> 'cat' | 'dog'
V -> 'sees' | 'likes'
""")

parser = nltk.ChartParser(grammar)

sentence = "the cat sees the dog".split()

for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```
