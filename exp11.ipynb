Simple Top-Down Parser

```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["Det", "N"]],
    "VP": [["V", "NP"]],
    "Det": [["the"]],
    "N": [["cat"], ["dog"]],
    "V": [["sees"], ["likes"]]
}


def parse(symbols, words):

    if not symbols:
        return len(words) == 0

    first = symbols[0]

    # Terminal symbol
    if first not in grammar:
        if words and words[0] == first:
            return parse(symbols[1:], words[1:])
        return False

    # Non-terminal
    for production in grammar[first]:
        if parse(production + symbols[1:], words):
            return True

    return False


sentence = "the cat sees the dog"
words = sentence.split()

if parse(["S"], words):
    print("Sentence is accepted by the grammar.")
else:
    print("Sentence is rejected by the grammar.")
```
