# EXPERIMENT 1: Context-Free Grammar Rule Validation
# and Parse Tree Construction

"""
QUESTION:

Design a Python program to validate whether a sentence follows
the given Context-Free Grammar (CFG) rules and generate its
parse tree representation.

Grammar:
S  → NP VP
NP → Det N
VP → V NP
Det → the | a
N   → dog | cat | bone
V   → chased | found

Tasks:
1. Accept a sentence as input.
2. Verify whether it conforms to the CFG.
3. Construct and return the parse tree.
4. Return "Invalid Sentence" if parsing fails.

Function Prototype:
def parse_cfg(sentence):
    pass
"""

CFG Program

```python
# CFG grammar definitions

grammar = {
    "Det": ["the", "a"],
    "N": ["dog", "cat", "bone"],
    "V": ["chased", "found"]
}


def parse_cfg(sentence):
    words = sentence.lower().split()

    # A valid sentence must contain exactly 5 words:
    # Determiner + Noun + Verb + Determiner + Noun
    if len(words) != 5:
        return "Invalid Sentence"

    det1, noun1, verb, det2, noun2 = words

    # Check grammar rules
    if det1 not in grammar["Det"]:
        return "Invalid Sentence"

    if noun1 not in grammar["N"]:
        return "Invalid Sentence"

    if verb not in grammar["V"]:
        return "Invalid Sentence"

    if det2 not in grammar["Det"]:
        return "Invalid Sentence"

    if noun2 not in grammar["N"]:
        return "Invalid Sentence"

    # Construct parse tree
    parse_tree = f"""
S
├── NP
│   ├── Det
│   │   └── {det1}
│   └── N
│       └── {noun1}
└── VP
    ├── V
    │   └── {verb}
    └── NP
        ├── Det
        │   └── {det2}
        └── N
            └── {noun2}
"""

    return "Valid Parse Tree:" + parse_tree
```

Test Cases

```python
# Test cases for Experiment 1

test_sentences = [
    "the dog chased a cat",
    "a cat found the bone",
    "dog chased cat",
    "the bone found a dog",
    "chased the dog bone"
]

for sentence in test_sentences:
    print("=" * 50)
    print("Input:", sentence)
    print(parse_cfg(sentence))
```

Individual Input

```python
# Accept a sentence from the user

sentence = input("Enter a sentence: ")

result = parse_cfg(sentence)

print("\nResult:")
print(result)
```
