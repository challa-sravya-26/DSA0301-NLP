Laplace (Add-One) Smoothing

```python
from collections import Counter

# Training corpus
text = "I love natural language processing I love python"

# Convert to lowercase and split into words
tokens = text.lower().split()

# Count unigrams and bigrams
unigrams = Counter(tokens)
bigrams = Counter(zip(tokens[:-1], tokens[1:]))

# Vocabulary size
V = len(set(tokens))

# Laplace smoothing function
def laplace_probability(prev_word, curr_word):
    bigram_count = bigrams[(prev_word, curr_word)]
    unigram_count = unigrams[prev_word]

    probability = (bigram_count + 1) / (unigram_count + V)
    return probability

# Example
print("Vocabulary:", set(tokens))
print("Vocabulary Size:", V)
print("Bigram Count (love, python):", bigrams[("love", "python")])
print("Laplace Probability:", laplace_probability("love", "python"))
```
