Probabilistic CFG Parsing

```python
import nltk

grammar = nltk.PCFG.fromstring("""
S -> NP VP [1.0]
NP -> Det N [0.6]
NP -> N [0.4]
VP -> V NP [1.0]
Det -> 'the' [1.0]
N -> 'cat' [0.5]
N -> 'dog' [0.5]
V -> 'sees' [1.0]
""")

parser = nltk.ViterbiParser(grammar)

sentence = "the cat sees the dog".split()

print("Probabilistic Parse Tree:")

try:
    for tree in parser.parse(sentence):
        print(tree)
        print("Probability:", tree.prob())
except ValueError:
    print("Sentence cannot be parsed.")
```
