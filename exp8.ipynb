Stochastic POS Tagging

```python
from collections import defaultdict, Counter

# Training data
training_data = [
    [("I", "PRON"), ("like", "VERB"), ("dogs", "NOUN")],
    [("I", "PRON"), ("like", "VERB"), ("cats", "NOUN")],
    [("dogs", "NOUN"), ("run", "VERB")],
    [("cats", "NOUN"), ("run", "VERB")],
    [("the", "DET"), ("dog", "NOUN"), ("runs", "VERB")]
]

# Count word-tag frequencies
word_tag_counts = defaultdict(Counter)

for sentence in training_data:
    for word, tag in sentence:
        word_tag_counts[word][tag] += 1

def stochastic_tagger(sentence):
    result = []

    for word in sentence.split():
        if word in word_tag_counts:
            tag = word_tag_counts[word].most_common(1)[0][0]
        else:
            tag = "NOUN"

        result.append((word, tag))

    return result


sentence = "I like dogs"

print("Stochastic POS Tags:")
print(stochastic_tagger(sentence))
```
