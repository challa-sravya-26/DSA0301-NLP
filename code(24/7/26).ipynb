Students implement Python programs using NLTK and collections.Counter to tokenize text, build a corpus, calculate word frequencies, generate unigram and bigram frequency distributions, compute probabilities, and visualize the results. give code for this

```python
!pip install nltk

import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.util import bigrams
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Sample text
text = """
Natural Language Processing is a branch of Artificial Intelligence.
Natural Language Processing enables computers to understand human language.
Language processing includes tokenization, stemming, parsing, and text generation.
"""

# Convert text to lowercase
text = text.lower()

# Tokenize the text
tokens = word_tokenize(text)

# Remove punctuation
words = [word for word in tokens if word.isalpha()]

print("Corpus (Tokens):")
print(words)

# ---------------------------------------------------
# Word Frequency using Counter
# ---------------------------------------------------
word_freq = Counter(words)

print("\nWord Frequencies:")
for word, freq in word_freq.items():
    print(f"{word}: {freq}")

# ---------------------------------------------------
# Unigram Frequency Distribution
# ---------------------------------------------------
print("\nUnigram Frequency Distribution")
for word, freq in word_freq.items():
    print(f"{word} -> {freq}")

# ---------------------------------------------------
# Bigram Frequency Distribution
# ---------------------------------------------------
bigram_list = list(bigrams(words))

bigram_freq = Counter(bigram_list)

print("\nBigram Frequency Distribution")
for bg, freq in bigram_freq.items():
    print(f"{bg} -> {freq}")

# ---------------------------------------------------
# Unigram Probabilities
# ---------------------------------------------------
total_words = len(words)

print("\nUnigram Probabilities")
for word, freq in word_freq.items():
    probability = freq / total_words
    print(f"P({word}) = {probability:.4f}")

# ---------------------------------------------------
# Bigram Probabilities
# ---------------------------------------------------
print("\nBigram Probabilities")

for bg, freq in bigram_freq.items():
    first_word = bg[0]
    probability = freq / word_freq[first_word]
    print(f"P({bg[1]} | {bg[0]}) = {probability:.4f}")

# ---------------------------------------------------
# Visualization
# ---------------------------------------------------
top_words = word_freq.most_common(10)

labels = [item[0] for item in top_words]
values = [item[1] for item in top_words]

plt.figure(figsize=(8,5))
plt.bar(labels, values)
plt.title("Top 10 Word Frequencies")
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.xticks(rotation=45)
plt.show()
```
