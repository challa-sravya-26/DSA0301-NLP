Finite-State Machine for Morphological Parsing

```python
# Experiment 4: Finite-State Machine for Plural Formation

def pluralize(word):
    # Rule 1: words ending in s, x, z, ch, sh
    if word.endswith(("s", "x", "z", "ch", "sh")):
        return word + "es"

    # Rule 2: consonant + y
    elif word.endswith("y") and len(word) > 1 and word[-2] not in "aeiou":
        return word[:-1] + "ies"

    # Rule 3: words ending in f or fe
    elif word.endswith("fe"):
        return word[:-2] + "ves"

    elif word.endswith("f"):
        return word[:-1] + "ves"

    # Default rule
    else:
        return word + "s"


words = ["cat", "dog", "bus", "box", "baby", "city", "knife", "leaf"]

print("Singular -> Plural")
print("-------------------")

for word in words:
    print(word, "->", pluralize(word))
```
