Word Sense Disambiguation Using Lesk Algorithm

```python
import nltk

nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn

def lesk(word, sentence):

    context = set(sentence.lower().split())

    best_sense = None
    max_overlap = 0

    for sense in wn.synsets(word):

        definition_words = set(
            sense.definition().lower().split()
        )

        overlap = len(context.intersection(definition_words))

        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = sense

    return best_sense


sentence = "I deposited money in the bank"

sense = lesk("bank", sentence)

if sense:
    print("Word:", "bank")
    print("Best Sense:", sense.name())
    print("Definition:", sense.definition())
else:
    print("Sense not found.")
```
