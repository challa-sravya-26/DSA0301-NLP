# EXPERIMENT 3: Probabilistic Context-Free Grammar (PCFG)
# Parsing

"""
QUESTION:

Implement a Python program to compute the probability of a
sentence using the given Probabilistic Context-Free Grammar.

Grammar:

S  → NP VP [1.0]
NP → Tom [0.7]
NP → Ann [0.3]
VP → sings [0.4]
VP → dances [0.6]

Probability Formula:

P(Sentence) =
P(S → NP VP) × P(NP) × P(VP)

Tasks:
1. Accept a two-word sentence.
2. Parse using the PCFG.
3. Compute sentence probability.
4. Return 0 if the sentence cannot be generated.

Function Prototype:
def sentence_probability(sentence):
    pass
"""

PCFG Rule

```python
# PCFG production probabilities

grammar_probabilities = {
    "S": {
        ("NP", "VP"): 1.0
    },

    "NP": {
        ("Tom",): 0.7,
        ("Ann",): 0.3
    },

    "VP": {
        ("sings",): 0.4,
        ("dances",): 0.6
    }
}
```

Probability Function

```python
def sentence_probability(sentence):
    # Remove extra spaces and split sentence
    words = sentence.strip().split()

    # Sentence must contain exactly two words
    if len(words) != 2:
        return 0.0

    subject = words[0].capitalize()
    verb = words[1].lower()

    # NP probabilities
    np_probability = {
        "Tom": 0.7,
        "Ann": 0.3
    }

    # VP probabilities
    vp_probability = {
        "sings": 0.4,
        "dances": 0.6
    }

    # Check whether words are present in grammar
    if subject not in np_probability:
        return 0.0

    if verb not in vp_probability:
        return 0.0

    # S -> NP VP probability
    s_probability = 1.0

    # Calculate sentence probability
    probability = (
        s_probability
        * np_probability[subject]
        * vp_probability[verb]
    )

    return probability
```

Test Cases

```python
# Test cases

test_sentences = [
    "Tom sings",
    "Tom dances",
    "Ann sings",
    "Ann dances",
    "Sam sings"
]

for sentence in test_sentences:
    probability = sentence_probability(sentence)
    print(f"{sentence} -> {probability:.2f}")
```

Individual Input

```python
sentence = input("Enter a two-word sentence: ")

probability = sentence_probability(sentence)

print("\nSentence:", sentence)
print("Probability:", f"{probability:.2f}")
```
