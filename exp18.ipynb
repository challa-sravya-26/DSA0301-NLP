Simple FOPC Parser

```python
import re

def parse_fopc(expression):

    expression = expression.strip()

    # Predicate pattern
    pattern = r'^[A-Za-z][A-Za-z0-9_]*\([A-Za-z0-9_, ]+\)$'

    if re.match(pattern, expression):
        print("Valid FOPC atomic expression")
        return

    # Negation
    if expression.startswith("NOT "):
        inner = expression[4:]
        print("Negation detected")
        parse_fopc(inner)
        return

    # AND
    if " AND " in expression:
        parts = expression.split(" AND ")
        print("AND operator detected")
        for part in parts:
            parse_fopc(part)
        return

    # OR
    if " OR " in expression:
        parts = expression.split(" OR ")
        print("OR operator detected")
        for part in parts:
            parse_fopc(part)
        return

    # Quantifiers
    if expression.startswith("FORALL "):
        print("Universal Quantifier detected")
        return

    if expression.startswith("EXISTS "):
        print("Existential Quantifier detected")
        return

    print("Invalid FOPC expression")


expressions = [
    "Student(John)",
    "Student(John) AND Smart(John)",
    "NOT Student(John)",
    "FORALL x Student(x)",
    "EXISTS x Student(x)"
]

for expression in expressions:

    print("\nExpression:", expression)
    parse_fopc(expression)
```
