Syntax-Driven Semantic Analysis

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')

from nltk import word_tokenize, pos_tag

sentence = "The intelligent student solved the difficult problem."

words = word_tokenize(sentence)
tags = pos_tag(words)

print("POS Tags:")
print(tags)

print("\nNoun Phrases:")

noun_phrases = []

current_np = []

for word, tag in tags:

    if tag in ["DT", "JJ", "NN", "NNS", "NNP", "NNPS"]:

        current_np.append(word)

    else:

        if current_np:
            noun_phrases.append(" ".join(current_np))
            current_np = []

if current_np:
    noun_phrases.append(" ".join(current_np))


for phrase in noun_phrases:

    print("Noun Phrase:", phrase)

    if "student" in phrase.lower():
        print("Meaning: Person who studies")

    elif "problem" in phrase.lower():
        print("Meaning: Question or task requiring a solution")

    else:
        print("Meaning: Entity or object")
```
